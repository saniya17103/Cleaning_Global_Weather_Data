## Cleanig the Global Weather Data 


```python
##Import Libraries
import pandas as pd
import numpy as np

df = pd.read_csv("Global_Weather_Repository.csv")
```

```python
##Check Data
print(df.shape)
df.head()
```

```python
## Check information
df.info()
```

```python
## Check Null Values
df.isnull().sum()
```

```python
## Check Duplicates
df.duplicated().sum()
```

```python

df.describe().T
```

```python
## Remove Duplicates
df = df.drop_duplicates().reset_index(drop=True)
```

```python
## Clean Column Names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
## Convert date time
df['last_updated'] = pd.to_datetime(
    df['last_updated'],
    errors='coerce'
)
```

```python
## Clean text columns
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python

df['wind_direction'] = df['wind_direction'].str.upper()
df['country'] = df['country'].str.title()
df['location_name'] = df['location_name'].str.title()
```

```python
## Validate geographical values
df.loc[
    (df['latitude'] < -90) | (df['latitude'] > 90),
    'latitude'
] = np.nan

df.loc[
    (df['longitude'] < -180) | (df['longitude'] > 180),
    'longitude'
] = np.nan
```

```python
## Clean Weather Values
df.loc[
    (df['humidity'] < 0) | (df['humidity'] > 100),
    'humidity'
] = np.nan
```

```python

df.loc[
    (df['cloud'] < 0) | (df['cloud'] > 100),
    'cloud'
] = np.nan
```

```python
df.loc[df['precip_mm'] < 0, 'precip_mm'] = np.nan
df.loc[df['precip_in'] < 0, 'precip_in'] = np.nan
```

```python
df.loc[
    (df['wind_degree'] < 0) | (df['wind_degree'] > 360),
    'wind_degree'
] = np.nan
```

```python
## Clean clearly invalid air-quality values
air_quality_columns = [
    'air_quality_carbon_monoxide',
    'air_quality_ozone',
    'air_quality_nitrogen_dioxide',
    'air_quality_sulphur_dioxide',
    'air_quality_pm2.5',
    'air_quality_pm10'
]

for col in air_quality_columns:
    df.loc[df[col] < 0, col] = np.nan
```

```python
## Clean Unrealistic Pressure
df.loc[
    (df['pressure_mb'] < 800) |
    (df['pressure_mb'] > 1100),
    'pressure_mb'
] = np.nan
```

```python
## Clean Unrealistic Wind Values
df.loc[df['wind_mph'] > 150, 'wind_mph'] = np.nan
df.loc[df['wind_kph'] > 240, 'wind_kph'] = np.nan

df.loc[df['gust_mph'] > 150, 'gust_mph'] = np.nan
df.loc[df['gust_kph'] > 240, 'gust_kph'] = np.nan
```

```python
## Validate air quality indexes
df.loc[
    (df['air_quality_us-epa-index'] < 1) |
    (df['air_quality_us-epa-index'] > 6),
    'air_quality_us-epa-index'
] = np.nan
```

```python
df.loc[
    (df['air_quality_gb-defra-index'] < 1) |
    (df['air_quality_gb-defra-index'] > 10),
    'air_quality_gb-defra-index'
] = np.nan
```

```python
Validate moon illumination
df.loc[
    (df['moon_illumination'] < 0) |
    (df['moon_illumination'] > 100),
    'moon_illumination'
] = np.nan
```

```python
## Handle missing values
df.isnull().sum().sort_values(ascending=False)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
## Cross Check
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("Missing values:")
print(df.isnull().sum().sum())
```

```python
print("Duplicate rows:")
print(df.duplicated().sum())
```
